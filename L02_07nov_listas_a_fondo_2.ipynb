# 📚 Parte 1 · La prueba de escritorio: ser el computador por un rato

Una **prueba de escritorio** es seguir el programa **a mano**, línea por línea, anotando en una tabla cuánto vale cada variable en cada momento. Usted hace de computador, con papel y lápiz.

**¿Para qué sirve?**
- Para **entender** un programa que escribió otra persona.
- Para **encontrar el error** cuando el programa corre pero da un resultado equivocado (ahí no hay mensaje rojo que ayude).
- Para **comprobar** su idea antes de escribirla.

### Cómo se hace
1. Haga una tabla con **una columna por cada variable** que cambia.
2. Escriba **una fila por cada vuelta** del ciclo.
3. En cada fila anote el valor con el que queda cada variable **al terminar esa vuelta**.
4. Lo que se imprime sale de la **última fila**.

### Ejemplo resuelto
```python
numeros = [4, 7, 2]
total = 0
for n in numeros:
    total = total + n
print(total)
```

Antes del ciclo: `total` vale `0`.

| Vuelta | `n` (el dato de esta vuelta) | Cuenta que hace | `total` al terminar la vuelta |
|---|---|---|---|
| 1 | 4 | 0 + 4 | 4 |
| 2 | 7 | 4 + 7 | 11 |
| 3 | 2 | 11 + 2 | 13 |

Se acabó la lista, el ciclo termina y `print(total)` muestra **13**.

> **Fíjese:** en cada vuelta, el `total` con el que se empieza es el `total` con el que terminó la vuelta anterior. Ese es todo el secreto de un acumulador.

Ejecute para comprobar que la tabla dijo la verdad:

````python
numeros = [4, 7, 2]
total = 0
for n in numeros:
    total = total + n
print(total)
````

# 📚 Parte 2 · Patrones 1 y 2: acumular y contar

Un **patrón** es una receta que se repite. Casi todos los problemas con listas se resuelven con **cinco** recetas. Cuando lea un problema, la pregunta es: *¿cuál de los cinco patrones es este?*

Los dos primeros ya los conoce de cuando trabajó con ciclos. Aquí están como receta, para que los tenga a la mano:

### Patrón 1 · Acumular («¿cuánto suma?»)
```python
total = 0                    # 1. ANTES del ciclo, en cero
for dato in lista:           # 2. recorrer
    total = total + dato     # 3. en cada vuelta se le suma EL DATO
print(total)                 # 4. se usa DESPUÉS del ciclo
```

### Patrón 2 · Contar («¿cuántos cumplen?»)
```python
cuantos = 0                  # 1. ANTES del ciclo, en cero
for dato in lista:           # 2. recorrer
    if dato > 100:           # 3. la condición que deben cumplir
        cuantos = cuantos + 1    # 4. se le suma UNO
print(cuantos)               # 5. se usa DESPUÉS del ciclo
```

**La diferencia que más se confunde:** el acumulador suma **el dato** (`+ dato`); el contador suma **uno** (`+ 1`).

**Lo nuevo de hoy:** los dos patrones se pueden **combinar en un solo recorrido**, y el acumulador también puede llevar condición («sume solo los que cumplan»):

````python
edades = [15, 22, 17, 30, 12]

suma_todas = 0          # acumulador sin condición
menores = 0             # contador con condición
suma_mayores = 0        # acumulador con condición

for edad in edades:
    suma_todas = suma_todas + edad
    if edad < 18:
        menores = menores + 1
    else:
        suma_mayores = suma_mayores + edad

print("Suma de todas las edades:", suma_todas)
print("Menores de edad:", menores)
print("Suma de las edades de los mayores:", suma_mayores)
````

**Qué pasó:** con **un solo** `for` se calcularon tres cosas. Cada variable se creó **antes** del ciclo y se mostró **después**.

### ⚠️ Los dos errores de siempre
| Error | Qué pasa | Arreglo |
|---|---|---|
| Crear el acumulador **dentro** del ciclo | En cada vuelta vuelve a quedar en 0 y al final solo tiene el último dato | La línea `total = 0` va **antes** del `for` |
| Poner el `print` **dentro** del ciclo | Imprime un resultado parcial en cada vuelta | El `print` va **después**, sin sangría |

# 📚 Parte 3 · Patrón 3: el mayor (o el menor) y **dónde está**

`max(lista)` dice **cuál** es el dato más grande, pero no dice **dónde** está. Y muchas veces lo que importa es el dónde: *¿qué día fue el de más ventas?*, *¿quién sacó la mejor nota?*

En el ejercicio 2 usted ya hizo la prueba de escritorio del patrón del mayor. Ahora le agregamos una segunda variable que recuerda **la posición**:

```python
mayor = lista[0]                 # 1. el primer dato es "el mayor hasta ahora"
pos_mayor = 0                    #    y está en la posición 0
for i in range(len(lista)):      # 2. recorrer las POSICIONES
    if lista[i] > mayor:         # 3. ¿este le gana al mayor hasta ahora?
        mayor = lista[i]         # 4. entonces es el nuevo mayor…
        pos_mayor = i            #    …y se anota dónde está
print(mayor, pos_mayor)          # 5. se usan DESPUÉS del ciclo
```

🏆 Es como el juego de «rey de la colina»: el primero se sube; cada uno que llega lo reta; si le gana, se queda arriba. Al final, el que está arriba es el mayor.

**Tres detalles que importan**
1. **Se recorre con `range(len(lista))`** porque necesitamos el número de la posición (`i`), no solo el dato.
2. **`mayor` empieza con `lista[0]`, no con 0.** Si empezara en 0 y todos los datos fueran negativos (temperaturas bajo cero, deudas), ninguno le ganaría al 0 y el resultado estaría mal.
3. **Para el menor** solo cambia el signo: `if lista[i] < menor:`.

````python
ventas = [120000, 85000, 150000, 97000]

mayor = ventas[0]
pos_mayor = 0
for i in range(len(ventas)):
    if ventas[i] > mayor:
        mayor = ventas[i]
        pos_mayor = i

print("La venta más alta fue", mayor)
print("Está en la posición", pos_mayor, "→ fue el día", pos_mayor + 1)
````

**Prueba de escritorio de ese ejemplo** (antes del ciclo: `mayor = 120000`, `pos_mayor = 0`):

| Vuelta | `i` | `ventas[i]` | ¿le gana a `mayor`? | `mayor` | `pos_mayor` |
|---|---|---|---|---|---|
| 1 | 0 | 120000 | no | 120000 | 0 |
| 2 | 1 | 85000 | no | 120000 | 0 |
| 3 | 2 | 150000 | **sí** | 150000 | 2 |
| 4 | 3 | 97000 | no | 150000 | 2 |

# 📚 Parte 4 · Patrón 4: buscar con bandera

`in` responde si un dato **exacto** está en la lista. Pero no sirve para preguntas como *«¿hay **alguna** nota menor que 3?»* o *«¿hay **algún** menor de edad?»*. Para eso se busca a mano, y la herramienta es la **bandera**.

Una **bandera** es una variable que solo vale `True` o `False` y que recuerda **si ya pasó algo**.

🚩 Es como el banderín del salvavidas en la playa: empieza abajo (`False`). Si alguien ve un tiburón, lo sube (`True`). Al final del día, con solo mirar el banderín se sabe si hubo tiburón.

```python
encontrado = False               # 1. ANTES del ciclo: todavía no lo he visto
for dato in lista:               # 2. recorrer
    if dato < 3.0:               # 3. ¿este es el que busco?
        encontrado = True        # 4. subo la bandera
        break                    # 5. ya lo encontré: no hace falta seguir
if encontrado:                   # 6. DESPUÉS del ciclo, miro la bandera
    print("Sí hay")
else:
    print("No hay")
```

### ⚠️ El error clásico: decir «no está» dentro del ciclo
Ejecute esta versión **equivocada** y cuente cuántas líneas imprime:

````python
codigos = [10, 20, 30, 40]
buscado = 30
for codigo in codigos:
    if codigo == buscado:
        print("Encontrado")
    else:
        print("No está")        # ⚠️ este else está en el lugar equivocado
````

Imprimió **cuatro** líneas y tres dicen «No está», aunque el 30 **sí** está. ¿Por qué? Porque ese `else` responde en **cada vuelta**: «este dato no es». Pero que **un** dato no sea el buscado no significa que **ninguno** lo sea.

> **Regla:** dentro del ciclo solo se puede saber que **sí** está. Para decir que **no** está hay que haber revisado **toda** la lista. Por eso el «no está» va **después** del ciclo, mirando la bandera.

# 📚 Parte 5 · Patrón 5: construir una lista nueva

Muchas veces el resultado no es un número sino **otra lista**, armada a partir de la primera. Hay dos sabores, y se pueden combinar:

| Sabor | Pregunta | La lista nueva tiene… |
|---|---|---|
| **Filtrar** | «¿cuáles cumplen?» | **menos** datos (o los mismos): solo los que pasan la condición |
| **Transformar** | «¿cómo queda cada uno después de…?» | **la misma cantidad** de datos, pero cada uno cambiado |

```python
nueva = []                           # 1. ANTES del ciclo, vacía
for dato in lista:                   # 2. recorrer
    if dato >= 3.0:                  # 3. (filtrar) ¿este entra?
        nueva.append(dato)           # 4. se agrega a la nueva
```

```python
nueva = []                           # 1. ANTES del ciclo, vacía
for dato in lista:                   # 2. recorrer
    nueva.append(dato * 2)           # 3. (transformar) se agrega el dato YA cambiado
```

El filtro ya lo usó (aprobados y reprobados). Mire ahora la **transformación**:

````python
precios = [1000, 2500, 800]

dobles = []
for precio in precios:
    dobles.append(precio * 2)       # se agrega el dato transformado

print("Originales:", precios)       # la lista original no cambia
print("Dobles:    ", dobles)
````

**Diferencia con lo de ayer:** ayer cambió los precios **dentro de la misma lista** con `precios[i] = ...`. Hoy la lista original queda intacta y el resultado va a una lista **nueva**. Se usa una u otra según si necesita conservar los datos originales.

# 📚 Parte 6 · Dos listas a la vez: listas paralelas

Hasta ahora cada lista guardaba **un solo tipo** de dato. Pero de una persona casi siempre interesan varios: su nombre **y** su nota. Una forma de guardarlos es con **dos listas emparejadas por la posición**:

```
nombres  =  ["Ana", "Luis", "Mía"]
notas    =  [ 4.5,   2.8,    3.9 ]
posición →    0       1       2
```

**La regla de oro:** lo que está en la **misma posición** de las dos listas pertenece a la **misma persona**. `nombres[1]` es Luis y `notas[1]` es la nota de Luis.

Para recorrerlas juntas se recorren las **posiciones**, y con la misma `i` se entra a las dos:

````python
nombres = ["Ana", "Luis", "Mía"]
notas = [4.5, 2.8, 3.9]

for i in range(len(nombres)):
    print(f"{i + 1}. {nombres[i]}: {notas[i]}")
````

### Las tres reglas para no dañar el emparejamiento
1. **Agregar siempre en las dos.** Si hace `nombres.append(...)`, en la línea siguiente va `notas.append(...)`.
2. **Quitar siempre en las dos, con la misma posición.** Primero se averigua la posición y después `nombres.pop(pos)` y `notas.pop(pos)`.
3. **No ordenar una sola.** Si ordena `notas` y deja `nombres` como estaba, cada persona queda con la nota de otra. Mírelo:

````python
nombres = ["Ana", "Luis", "Mía"]
notas = [4.5, 2.8, 3.9]

notas.sort()                         # ⚠️ se ordenó solo una de las dos
for i in range(len(nombres)):
    print(nombres[i], notas[i])      # Ana ya no tiene 4.5: se dañó el emparejamiento
````

### El poder de combinar patrones
Con el patrón del mayor usted encuentra **la posición** de la nota más alta. Y como las listas están emparejadas, esa misma posición le dice **de quién es**: `nombres[pos_mayor]`.
